```python
from pyprojroot import here
PROJECT_ROOT = here()
# IR: inpulse response
IR_DIR = PROJECT_ROOT / "Pz and Sz"
FIG_DIR = PROJECT_ROOT / "figs" / "impulse_response"
print(PROJECT_ROOT)
print(IR_DIR)
print(FIG_DIR)
FIG_DIR.mkdir(parents=True, exist_ok=True)
```

```python
import matplotlib.pyplot as plt
from gfanc.common.Reading_path_test import loading_paths_from_MAT
import numpy as np
```

```python
subfolder = "Dongyuan"
Pri_path_file_name, Sec_path_file_name = "Primary_path.mat", "Secondary_path.mat"
Pri_path, Sec_path = loading_paths_from_MAT(IR_DIR, subfolder, Pri_path_file_name, Sec_path_file_name)
print(np.argmax(Pri_path), np.argmax(Sec_path))
plt.rcParams.update({
    "font.size": 12,
    "axes.labelsize": 14,
    "axes.titlesize": 14,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11,
    "legend.fontsize": 11,
    "figure.figsize": (8, 4),
    "axes.grid": True,
    "grid.alpha": 0.3,
})
def plot_ir(path):
    fig, ax = plt.subplots()

    ax.plot(
        np.arange(len(path)),
        path,
        linewidth=1.5,
        color="red"
    )

    ax.set_xlabel("Sample")
    ax.set_ylabel("Amplitude")
    ax.set_title("Impulse Response")

    ax.grid(True, alpha=0.3)
    ax.set_xlim(0, len(path) - 1)

    plt.tight_layout()
    plt.savefig(FIG_DIR/subfolder)
    plt.show()
```

```python
fig, ax = plt.subplots()

ax.plot(
    np.arange(len(Pri_path)),
    Pri_path,
    linewidth=1.5,
    color="red",
    label="Primary_path",
)

ax.plot(
    np.arange(len(Sec_path)),
    Sec_path,
    linewidth=1.5,
    color="blue",
    label="Secondary_path",
)

ax.set_xlabel("Sample")
ax.set_ylabel("Amplitude")
ax.set_title("Impulse Response")

ax.grid(True, alpha=0.3)
ax.set_xlim(0, len(Pri_path) - 1)
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR/subfolder)
plt.show()
```
