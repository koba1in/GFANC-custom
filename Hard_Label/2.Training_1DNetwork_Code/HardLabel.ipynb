```python
import pandas as pd
path = "Original_Dataset/Synthesized_Dataset/Validate_data/"
```

```python
df = pd.read_csv(path+"Soft_Index.csv")
import ast
df["Class_ID"] = df["Class_ID"].apply(ast.literal_eval)
df["Class_ID"]
```

```python
def soft_to_hard(soft_array):
    hard_array = []
    for i in soft_array:
        if i < 1/3:
            hard_array.append(0)
        elif i < 2/3:
            hard_array.append(1)
        else:
            hard_array.append(2)
    return hard_array
df["Class_ID"] = df["Class_ID"].apply(soft_to_hard)
df.to_csv(path+"Hard_Index_3.csv", index=False)
```

```python
import pandas as pd
path = "Original_Dataset/Synthesized_Dataset/Training_data/Hard_Index_3.csv"
df = pd.read_csv(path)
```

```python
import ast
id = df["Class_ID"].apply(ast.literal_eval)
```

```python
zeros = id.apply(lambda x: sum([i == 0 for i in x])).sum()
ones = id.apply(lambda x: sum([i == 1 for i in x])).sum()
tows = id.apply(lambda x: sum([i == 2 for i in x])).sum()
total = zeros + ones + tows
```

```python
print(zeros/total, ones/total, tows/total)
```
