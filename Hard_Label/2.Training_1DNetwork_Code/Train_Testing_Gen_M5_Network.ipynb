```python
from training.Train_validate import Train_Validate_CNN, General_Train_Validate_CNN
from training.Testing_M5_Network_original import Test_model_accuracy_original, Output_Test_Error_Samples
from training.Bcolors import bcolors
import os
import torch
import matplotlib.pyplot as plt
import numpy as np
 
print("number of GPUs:", torch.cuda.device_count())
```

```python
# Training and validating a new CNN model
BaseDir = "Original_Dataset/"
Folder_name_list_of_data_set=['Synthesized_Dataset/Training_data', 'Synthesized_Dataset/Validate_data', 'Synthesized_Dataset/Testing_data']
Folder_name_list_of_data_set = [BaseDir+i+"/" for i in Folder_name_list_of_data_set]
MODEL_PTH = 'models_from_original_dataset/M6_res_Synthetic'
File_sheet = 'Hard_Index'

acc_train, acc_validate, train_loss_epochs, validate_loss_epochs = General_Train_Validate_CNN(TRIAN_DATASET_FILE=Folder_name_list_of_data_set[0]
                                                        , VALIDATION_DATASET_FILE=Folder_name_list_of_data_set[1]
                                                        , MODEL_PTH=MODEL_PTH, File_sheet=File_sheet,num=None,numclass=3)
plt.plot(np.arange(len(train_loss_epochs)),train_loss_epochs, marker='.', markersize=10, color='r',label='training_loss')
plt.plot(np.arange(len(validate_loss_epochs)),validate_loss_epochs, marker='.', markersize=10, color='g',label='validation_loss')
plt.ylabel('Loss')
plt.xlabel('Epoches')
plt.legend()
plt.savefig('Training_loss_' + '.png', dpi=600, bbox_inches='tight', pad_inches=0)
plt.show()
```
