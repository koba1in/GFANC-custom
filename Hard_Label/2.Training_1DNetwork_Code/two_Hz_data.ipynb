```python
import numpy as np 
import pandas as pd 
old_training_data_dir = "Original_Dataset/Synthesized_Dataset/Training_data/"
old_validate_data_dir = "Original_Dataset/Synthesized_Dataset/Validate_data/"
new_training_data_dir = "Original_Dataset/Synthesized_Dataset/New_Training_data/"
new_validate_data_dir = "Original_Dataset/Synthesized_Dataset/New_Validate_data/"
Hard_Index = "Hard_Index.csv"
```

```python
df = pd.read_csv(old_training_data_dir + Hard_Index)
max = df.shape[0]
num = max // 2
rations = 0.5 + 0.5 * np.random.random(num)
mix_index = np.random.randint(0, max, num)
mix_data = df.loc[mix_index]
base_data = df[-num:]
import ast
mix_data_Class_ID = mix_data["Class_ID"].apply(ast.literal_eval).apply(np.array).values
base_data_Class_ID = base_data["Class_ID"].apply(ast.literal_eval).apply(np.array)
base_data_Class_ID = (base_data_Class_ID + mix_data_Class_ID).apply(lambda x: [ 0 if i == 0 else 1 for i in x ]).apply(str)
df.update(base_data_Class_ID)
df.to_csv(new_training_data_dir + Hard_Index, index=False)
```

```python
mix_filename = df.loc[mix_index]["File_path"].values
import torchaudio
import torch
base_dir = old_training_data_dir
for i in range(num):
    if i % 1000 == 0:
        print(i)
    base, fs = torchaudio.load(base_dir + base_data["File_path"].values[i])
    mix, fs = torchaudio.load(base_dir + mix_filename[i])
    wave = rations[i] * base.numpy() + (1 - rations[i]) * mix.numpy()
    torchaudio.save(new_training_data_dir + base_data["File_path"].values[i], torch.tensor(wave), fs)
```
