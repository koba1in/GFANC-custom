## Secondary pathに遅延を追加する

`def_delay_secondary_path` は、主経路（primary path）と副経路（secondary path）のインパルス応答を読み込み、副経路の先頭にゼロを追加して遅延させます。遅延後の主経路・副経路は、それぞれ元と同じファイル名で `savefolder` に保存されます。

### 引数

- `folder`: 入力MATファイルがあるフォルダ
- `pri_path_file`, `sec_path_file`: 主経路・副経路のMATファイル名
- `pri_path_name`, `sec_path_name`: 各MATファイル内の変数名
- `savefolder`: 処理結果の保存先フォルダ
- `delay`: 追加する遅延サンプル数（省略時は `0`）

主経路と副経路それぞれの最大振幅の位置を比較し、副経路の最大振幅が主経路と同じ位置になるようにゼロを追加します。さらに `delay` を指定すると、その分だけ副経路を追加で遅らせます。

### 使用例

```python
def_delay_secondary_path(
    folder="input",
    pri_path_file="primary.mat",
    sec_path_file="secondary.mat",
    pri_path_name="pri_path",
    sec_path_name="sec_path",
    savefolder="output",
    delay=0,
)

````python
delays = [-25, 0]
pri_path_file = "Primary_path.mat"
sec_path_file = "Secondary_path.mat"
pri_path_name = "Pz1"
sec_path_name = "S"
for delay in delays:
    save_folder = IR_DIR / ("Secondary_path_delay_" + str(delay) + "samples")
    Path.mkdir(save_folder, parents=True, exist_ok=True)
    delay_secondary_path(IR_DIR/"Dongyuan", pri_path_file, sec_path_file, pri_path_name, sec_path_name, save_folder, delay)
````
