```python
num = 0
result_sum = 0
for i in results_original:
    num += len(i)
    
    result_sum += torch.sum(torch.pow(i, torch.tensor(2)))
original = result_sum / num
num = 0
result_sum = 0
for i in results_new:
    num += len(i)
    result_sum += torch.sum(torch.pow(i, torch.tensor(2)))
new = result_sum / num
```

```python
Leq_original = 10 * np.log10(original.numpy())
Leq_new = 10 * np.log10(new.numpy())
print("improvement = ", Leq_original - Leq_new, "dB")
```
